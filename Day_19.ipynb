```python
#infix to prefix conversion using stacks
def precedence(op):
  if op=="+" or op=="-":
    return 1
  if op=="*" or op=="/":
    return 2
  if op=="^":
    return 3
def infoixtoprefix(expression):
  stack=[]
  output=[]
  expression=expression[::-1]
  for ch in expression:
    if ch.isalnum():
      output.append(ch)
    else:
      while stack and precedence(ch)<precedence(stack[-1]):
        output.append(stack.pop())
        stack.append(ch)
  while stack:
    output.append(stack.pop())
  return ''.join(output[::-1])
expression=input("Enter expression")
prefix=infoixtoprefix(expression)
print(prefix)
```

```python
#infix to postfix conversion using stacks
def precedence(op):
  if op=="+" or op=="-":
    return 1
  if op=="*" or op=="/":
    return 2
  if op=="^":
    return 3
def infixtopostfix(expression):
  stack=[]
  output=[]
  for ch in expression:
    if ch.isalnum():
      output.append(ch)
    else:
      while stack and precedence(ch)<precedence(stack[-1]):
        output.append(stack.pop())
        stack.append(ch)
  while stack:
    output.append(stack.pop())
  return ''.join(output)
expression=input("Enter expression")
postfix=infixtopostfix(expression)
print(postfix)
```

```python
#prefix to postfix using stacks
#prefix to postfix using stacks
def prefixtopostfix(expression):
    stack = []
    for ch in expression[::-1]:
        if ch.isalnum():
            stack.append(ch)
        else:
            operand1 = stack.pop()
            operand2 = stack.pop()
            result = operand1 + operand2 + ch
            stack.append(result)
    return stack.pop()
expression = input("Enter expression: ")
postfix = prefixtopostfix(expression)
print("Postfix expression:", postfix)
```
