```python
#priority queue-hospital management system
import heapq
hospital=[]
n=int(input("enter number of patients:"))
for i in range(n):
  name=input("enter patient name:")
  purpose=input("enter purpose(consultation/follow-up/Emergency):")
  if purpose=='Emergency':
    priority=1
  elif purpose=='Consultation':
    priority=2
  else:
    priority=3
  heapq.heappush(hospital,(priority,name,purpose))
print("\n Patients treated in order:")
while hospital:
  priority,name,purpose=heapq.heappop(hospital)
  print(name,"-",purpose)
```

```python
#priority queue-using package
from queue import PriorityQueue
pq=PriorityQueue()
n=int(input("Enter number of elements:"))
for i in range(n):
  value=input("Enter value")
  priority=int(input("enter priority:"))
  pq.put((priority,value))
print("\n Priority queue:")
while not pq.empty():
  priority,value=pq.get()
  print("Value",value,"Priority:",priority)
```

```python
#priority queue-using lists
queue=[]
n=int(input("Enter number of elements:"))
for i in range(n):
  value=input("enter value:")
  priority=int(input("enter priority:"))
queue.append((priority,value))
queue.sort()
print("\n Priority Queue:")
for priority,value in queue:
  print("Value",value,"priority:",priority)
priority,value=queue.pop(0)
print("\nDeleted element:")
print("Value:",value,"priority:",priority)
print("\nRemaining priority queue:")
for priority,value in queue:
  print("value",value,"priority:",priority)
```
