```python
#find the pair with given sum using 2 pointers approach
arr=list(map(int,input("Enter elements").split()))
target=int(input("Enter target"))
left=0
right=len(arr)-1
found=False
while left<right:
  total=arr[left]+arr[right]
  if total==target:
    print("pair found values: ", arr[left],arr[right])
    print("pair found index: ",left,right)
    found=True
    break
  elif total<target:
    left+=1
  else:
    right-=1
if not found:
  print("No pair found ")
```

```python
#remove the duplicate elements from a sorted bytearray
arr=list(map(int,input("enter elements:").split()))
slow=0
fast=0
for fast in range(1,len(arr)):
  if arr[fast]!=arr[slow]:
    slow+=1
    arr[slow]=arr[fast]
print(arr[:slow+1])
```
