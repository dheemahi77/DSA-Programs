```python
#1.constant time O(1)#static value
n=int(input("enter a number:"))
print("square of the number:",n*n)
```

```python
#2.linear time O(n)
n=int(input("enter the number:"))
for i in range(1,n+1):
  print(i)
```

```python
#3.quadratic time O(n^2)
n=int(input("enter the number:"))
for i in range(1,n+1):
  for j in range(1,n+1):
    print(i,end=' ')
  print()
```

```python
#4.logarithmic time O(log n)
n=int(input("enter the number:"))
while n>1:
  print(n)
  n//=2

```

```python
#5.independent time O(2n)
n=int(input("enter the number:"))
for i in range(n):
  print(i,end=' ')
for j in range(n):
  print(j,end=" ")
```

```python
#6.best time O(n//2)
n=int(input("enter the number:"))
for i in range(n//2):
  print(i,end=' ')
```

```python
#7.constant+linear time O(n+2)
n=int(input("enter the number:"))
for i in range(n):
  print(i,end=' ')
print('hi')
print("students")
```

```python
#8.exponential time O(2^n)
def abc(n):
  if n==0:
    return
  print(n)
  abc(n-1)
  abc(n-1)
n=int(input("enter a number:"))
abc(n)
```
