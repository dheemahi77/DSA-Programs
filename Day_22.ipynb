```python
#Dequeue
from collections import deque
dq=deque([1,2,3,4])
dq.append(5)
print(dq)
dq.appendleft(0)
print(dq)
print(dq.pop())
print(dq.popleft())
print(dq)
dq.extend([6,7,8])
print(dq)
dq.extendleft([1,2,3])
print(dq)
dq.rotate(2)
print(dq)
dq.rotate(-4)
print(dq)
dq.rotate(2)
print(dq)
dq.remove(7)
print(dq)
print(dq.count(1))
print(dq.index(4))
dq.reverse()
print(dq)
dq.clear()
print(dq)
```
