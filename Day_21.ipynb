```python
#insertion in a queue operation
from queue import Queue
q=Queue()
print("===Queue=Enqueue operation===")
print("Enter elements-to finish enterinf type stop:")
while True:
    item=input("Enter element to Enqueue:")
    if item.lower()=='stop':
        break
    q.put(item)
    print(f'{item} has been inserted in to Queue')
    print(f"Queue: {list(q.queue)}\n")
print("\n Final Queue:", list(q.queue()))
print("Total items:",q.qsize)
```

```python
#insertion in a queue operation
from queue import Queue
q=Queue()
print("===Queue-Enqueue operation===")
print("Enter elements-to finish enterinf type stop:")
while True:
    item=input("Entestopr element to Enqueue:")
    if item.lower()=='stop':
        break
    q.put(item)
    print(f'{item} has been inserted in to Queue')
    print(f"Queue: {list(q.queue)}\n")
print("\n Final Queue:", list(q.queue()))
print("Total items:",q.qsize)
while not q.empty():
    confirm=input("Press enter to dequeue the element:")
    if confirm.lower()=="stop":
        break
    item=q.get()
    print("Dequeued:",item)
    print("\nRemaining queue:",list(q.queue))
if q.empty():
    print("Queue is empty")
else:
    print("Final queue:",list(q.queue))
```

```python
#insertion in a queue operation
from queue import Queue
q=Queue(maxsize=4)
print("===Queue-Enqueue operation===")
print("Enter elements-to finish type stop:")
print("Maximum size:",q.maxsize)
print("is the queue initially full",q.full())
while True:
    if q.full():
        print("Queue overflow...")
        print("Current queue:",list(q.queue))
        break
    item=input("Enter element to Enqueue:")
    if item.lower()=='stop':
        break
    q.put(item)
    print(f'{item} has been inserted in to Queue')
    print(f"Queue: {list(q.queue)}\n")
print("\n Final Queue:", list(q.queue))
print("Total items:",q.qsize())
print("Peek element:",q.queue[0])
```
