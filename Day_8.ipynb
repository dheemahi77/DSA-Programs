**Searching and manipulations in arrays**

```python
#checking if an element is present in the array or not
arr=[]
n=int(input("Enter a size of an array:"))
print("Enter elements:")
for i in range(n):
  arr.append(int(input()))
target=int(input("enter the element you want to find:"))
if target in arr:
  print(target,"Found")
else:
  print(target,"Not found")
```

```python
#checking if an element is index in the array or not
arr=[]
n=int(input("Enter a size of an array:"))
print("Enter elements:")
for i in range(n):
  arr.append(int(input()))
target=int(input("enter the element you want to find:"))
if target in arr:
  print(target,"Found",arr.index(target))
else:
  print(target,"Not found")
```

```python
#checking if an element is present in the array at index position
arr=[]
n=int(input("Enter a size of an array:"))
print("Enter elements:")
for i in range(n):
  arr.append(int(input()))
target=int(input("enter the element you want to find:"))
found=False
for i in range(n):
  if arr[i]==target:
    print(target,"Found at index:",i)
    found=True
    break
if found==False:
    print(target,"Not found")
```

```python
#count how many items the element is present in the array
c=0
arr=[]
n=int(input("Enter a size of an array:"))
print("Enter elements:")
for i in range(n):
  arr.append(int(input()))
target=int(input("enter the element you want to find:"))
for i in range(n):
  if arr[i]==target:
    c+=1
print("count of",target,"is",c)
```
