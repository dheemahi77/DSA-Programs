```python
import heapq
heap=[]
n=int(input("enter number if elements: "))
for i in range(n):
  value=int(input("enter element"))
  heapq.heappush(heap,value)
print("min heap",heap)
x=int(input("Enter value to push:"))
removed=heapq.heappushpop(heap,x)
print("Removed element:",removed)
print("Min heap:",heap)
```

```python
#using replace method
import heapq
heap=[]
n=int(input("enter a number of elements:"))
for i in range(n):
  value=int(input("enter element:"))
  heapq.heappush(heap,value)
print("min heap",heap)
x=int(input("enter a value to push"))
removed=heapq.heapreplace(heap,x)
print("removed element:",removed,"by",x)
```

```python
#using largest and smallest method
import heapq
heap=[]
n=int(input("enter number of elements:"))
for i in range(n):
  value=int(input("enter element:"))
  heapq.heappush(heap,value)
print("min heap",heap)
x=int(input("enter a value to push"))
small=heapq.nsmallest(x,heap)
print("smallest elements:",small)
large=heapq.nlargest(x,heap)
print("largest elements:",large)
```
