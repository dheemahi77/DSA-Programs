```python
#Increasing monotonic stack
arr=list(map(int,input("Enter values:").split()))
stack=[]
for value in arr:
  while stack and stack[-1]>value:
    stack.pop()
  stack.append(value)
print(*stack)
```

```python
#Decresing monotonic stack
arr=list(map(int,input("Enter values:").split()))
stack=[]
for value in arr:
  while stack and stack[-1]<value:
    stack.pop()
  stack.append(value)
print(*stack)
```

```python
#Bitonic stack / 1 3 5 4 2-peak
arr=list(map(int,input("Enter values:").split()))
left=[]
right=[]
for i in range(len(arr)):
  if i%2==0:
    left.append(arr[i])
  else:
    right.append(arr[i])
right.reverse()
result=left+right
print(*result)
```

```python
#Bitonic stack / 5 3 1 2 4-vally
arr=list(map(int,input("Enter values:").split()))
arr.sort()
result=[]
#odd positions in reverse order
for i in range(len(arr)-1,-1,-2):
  result.append(arr[i])
#even positions normal order
for i in range(len(arr)%2,len(arr),2):
  result.append(arr[i])
print(result)
```
