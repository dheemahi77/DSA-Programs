```python
#Reverse an array
n=int(input("enter a number of elements:"))
arr=[]
for i in range(n):
  val=int(input("enter a number:"))
  arr.append(val)
print("original array:")
for i in arr:
  print(i,end=" ")
print("Reversed array:")
i=n-1
while i>=0:
  print(arr[i],end=" ")
  i-=1
```

```python
#code to rotate an array to left
n=int(input("enter number of elements:"))
arr=[]
for i in range(n):
  val=int(input("enter a value:"))
  arr.append(val)
print("original array:")
for i in arr:
  print(i,end=" ")
print()
print("rotated to left array:")
first=arr[0]
for i in range(0,n-1):
  arr[i]=arr[i+1]
arr[n-1]=first
for i in range(n):
  print(arr[i],end=' ')
```

```python
#code to rotate an array to right1
n=int(input("Enter no.of elements: "))
arr=[]
for i in range(n):
  val=int(input("Enter a values:"))
  arr.append(val)
print("Original array: ")
for i in arr:
  print(i,end=' ')
print()
print("Rotated to right array: ")
if n > 0:
  last=arr[n-1]
  for i in range(n-1, 0, -1):
    arr[i]=arr[i-1]
  arr[0]=last
for i in range(n):
  print(arr[i],end=' ')
```

```python
#code to insert an element in an array with respect to position value
n=int(input("Enter number of elements: "))
arr=[]
for i in range(n):
  val=int(input("Enter a values:"))
  arr.append(val)
print("Original array: ")
for i in arr:
  print(i,end=' ')
print()
pos=int(input("enter the position:"))
value=int(input("enter the value of position:"))
arr.insert(pos,value)
print("Latest array after inserting an element:")
for i in arr:
  print(i,end=" ")
print()
```

```python
#code to delete an element in an array with respect to position value
n=int(input("Enter number of elements: "))
arr=[]
for i in range(n):
  val=int(input("Enter a values:"))
  arr.append(val)
print("Original array: ")
for i in arr:
  print(i,end=' ')
print()
pos=int(input("enter the position:"))
arr.pop(pos)
print("Delete an element:")
for i in arr:
  print(i,end=" ")
print()
```

```python
#code to update an element in an array with respect to position value
n=int(input("Enter number of elements: "))
arr=[]
for i in range(n):
  val=int(input("Enter a values:"))
  arr.append(val)
print("Original array: ")
for i in arr:
  print(i,end=' ')
print()
pos=int(input("enter the position:"))
value=int(input("enter the value position:"))
arr[pos]=value
print("Latest array modify an element:")
for i in arr:
  print(i,end=" ")
print()
```
