**Binary Search Variations**

```python
#Find a square root of a number using binary search pattern
n=int(input("enter a value:"))
left=0
right=n
ans=0
while left<=right:
  mid=(left+right)//2
  if mid*mid==n:
    ans=mid
    break
  elif mid*mid<n:
    ans=mid
    left=mid+1
  else:
    right=mid-1
print("Square root:",ans)
```

```python
#2. find the first occurence of target value using binary search pattern
# -1 0 1 2 2 2 3 4
# target=2
arr=list(map(int,input("enter elements :").split()))
left=0
right=len(arr)-1
target=int(input("enter target:"))
ans=-1
while left<=right:
  mid=(left+right)//2
  if arr[mid]==target:
    ans=mid
    right=mid-1
  elif arr[mid]<target:
    left=mid+1
  else:
    right=mid-1
print("first occurence index of :",target,"=",ans)
```

```python
# 3. find the last occurence of a number using binary search
arr=list(map(int,input("enter elements :").split()))
left=0
right=len(arr)-1
target=int(input("enter target:"))
ans=-1
while left<=right:
  mid=(left+right)//2
  if arr[mid]==target:
    ans=mid
    left=mid+1
  elif arr[mid]<target:
    left=mid+1
  else:
    right=mid-1
print("first occurence index of :",target,"=",ans)
```

```python
# 4. find the count of occurence of a number using binary search pattern
arr=list(map(int,input("enter elements :").split()))
left=0
right=len(arr)-1
target=int(input("enter target:"))
first=-1
while left<=right:
  mid=(left+right)//2
  if arr[mid]==target:
    first=mid
    right=mid-1
  elif arr[mid]<target:
    left=mid+1
  else:
    right=mid-1
print("first occurence index of :",target,"=",first)
# last occurence
left=0
right=len(arr)-1
last=-1
while left<=right:
  mid=(left+right)//2
  if arr[mid]==target:
    last=mid
    left=mid+1
  elif arr[mid]<target:
    right=mid+1
  else:
    right=mid-1
print("first occurence index of :",target,"=",last)
if first==-1:
  print("count:",0)
else:
  print("count:", last-first+1)
```
