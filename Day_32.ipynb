```python
# n- Queens by back tracking
n= int(input("enter n: "))
board = [-1] *n
count= 0
def is_safe(row, col):
    for i in range(row):
        #same column
        if board[i]==col:
            return False
        #same diagnol
        if abs(board[i]- col)==abs(i-row):
            return False
    return True
def backtracking(row):
    global count
    # all queens placed
    if row==n:
        count+=1
        print("Solution", count)
        for i in range(n):
            for j in range(n):
                if board[i]==j:
                    print("Q", end= ' ')
                else:
                    print("_", end=' ')
            print()
        print()
        return
    # try every column
    for col in range(n):
        if is_safe(row,col):
            #place queen
            board[row]= col
            #next row
            backtracking(row+1)
            #backtrack
            board[row]= -1
backtracking(0)
print("Total Solutions ",count )
```

```python
#maze by back tracking
maze=[]
print("enter 4x4 maze(1=open,0=close)")
for i in range(4):
  row=list(map(int,input().split()))
  maze.append(row)
path=[]
def way(row,col):
  if row==3 and col==3:
    print("path:","".join(path))
    return True
  if row<0 or row>=4 or col<0 or col>=4:
    return False
  if maze[row][col]==0:
    return False
  maze[row][col]=0
  path.append("D")
  if way(row+1,col):
    return True
  path.pop()
  path.append("R")
  if way(row,col+1):
    return True
  path.pop()
  return False
if way(0,0)==False:
  print("no path")
```
