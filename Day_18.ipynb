```python
# What is a Stack?

A **stack** is a linear data structure that follows a particular order in which operations are performed. The order may be **LIFO (Last In First Out)** or **FILO (First In Last Out)**. This means the last element added to the stack is the first one to be removed.

Think of it like a stack of plates: you can only add a new plate on top, and you can only remove the top-most plate.

# Key Stack Operations:
Push**: Adds an item to the top of the stack.
Pop**: Removes the top item from the stack.
Peek/Top**: Returns the top item without removing it.
isEmpty**: Checks if the stack is empty.
isFull**: Checks if the stack is full (relevant for fixed-size stacks).

Stacks are commonly used in various computing scenarios, such as function call management, expression evaluation, and undo/redo functionalities.
```

```python
#stack push operation
stack=[]
size=int(input("Enter stack size:"))
n=int(input("Enter number of elements you want to push:"))
for i in range(n):
  if len(stack)<size:
    val=int(input("Enter value:"))
    stack.append(val)
  else:
    print("stack over flow ...")
print("stack:",stack)
print("peek element: ",stack[-1])
print("removed from stak:",stack.pop())
print("stack:",stack)
print("peek element:",stack[-1])
```

```python
#stack empty operation
stack=[]
size=int(input("Enter stack size:"))
n=int(input("Enter number of elements you want to push:"))
for i in range(n):
  if len(stack)<size:
    val=int(input("Enter value:"))
    stack.append(val)
  else:
    print("stack over flow ...")
print("stack:",stack)
while True:
  if len(stack)==0:
    print("stack empty")
    break
  else:
    print("popped:",stack.pop())
    print("latest stack",stack)
```
