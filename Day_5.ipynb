**Basic Approaches**

```python
#1.math approach
n=int(input("enter a value:"))
if n%2==1:
  print(n,'is odd')
else:
  print(n,"is even")
```

```python
#2.naive approach
arr=list(map(int,input("enter element").split()))
largest=arr[0]
for i in range(len(arr)):
  if arr[i]>largest:
    largest=arr[i]
print("largest:",largest)
```

```python
#brute-force approach
arr=list(map(int,input("enter element").split()))
target=int(input("enter target:"))
found=False
for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i]+arr[j]==target:
      print("pair found at:",i,j)
      print("pair founf at:",arr[i],arr[j])
      found=True
      break
  if found:
    break
if not found:
  print("No pair found")
```

```python
#greedy approach
coins=list(map(int,input("enter elements").split()))
amount=int(input("enter target:"))
print("coins used:")
for coin in coins:
  while amount>=coin:
    print(coin,end=" ")
    amount-=coin
```

```python
#back tracking
def generate(s,n):
  if len(s)==n:
    print(s)
    return
  generate(s+'0',n)
  generate(s+'1',n)
n=int(input("enter length:"))
generate("",n)
```
