```python
#permutations of a string
text=input("enter text:")
n=len(text)
for i in range(n):
  for j in range(n):
    if j==i:
      continue
    for k in range(n):
      if k==i or k==j:
        continue
      print(text[i]+text[j]+text[k])
```

```python
#permutations of a string using recursion
def permute(s,ans):
  if len(s)==0:
    print(ans,end=" ")
    return
  for i in range(len(s)):
    bal=s[:i]+s[i+1:]
    permute(bal,ans+s[i])
s=input("enter text:")
permute(s,"")
```

```python
#combinations of a string
s=input("enter a string:")
n=len(s)
for i in range(n):
  for j in range(i+1,n):
    print(s[i]+s[j])
```

```python
#combinations probability of a string
def combination(s,start,ans):
  if ans!="":
    print(ans)
  for i in range(start,len(s)):
    combination(s,i+1,ans+s[i])
s=input("enter string:")
combination(s,0,"")
```
