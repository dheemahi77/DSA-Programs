```python
#Heap Queue using max heap
arr=list(map(int,input("Enter elements:").split()))
for i in range(len(arr)):
  child=i
  while child>0:
    parent=(child-1)//2
    if arr[child]>arr[parent]:
      arr[child],arr[parent]=arr[parent],arr[child]
      child=parent
    else:
      break
print("Max Heap",arr)
```
