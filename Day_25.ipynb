```python
#create a  #pre order
class node:
  def __init__(self,data):
    self.data=data
    self.left=None
    self.right=None
values=input("enter elements:")
root=node(values[0])
queue=[root]
i=1
while i<len(values):
  current=queue.pop(0)
  if i<len(values):
    current.left=node(values[i])
    queue.append(current.left)
    i+=1
  if i<len(values):
    current.right=node(values[i])
    queue.append(current.right)
    i+=1
print("Tree Created")
```

```python
#create a tree #inorder
class node:
  def __init__(self,data):
    self.data=data
    self.left=None
    self.right=None
values=input("enter elements:")
root=node(values[0])
queue=[root]
i=1
while i<len(values):
  current=queue.pop(0)
  if i<len(values):
    current.left=node(values[i])
    queue.append(current.left)
    i+=1
  if i<len(values):
    current.right=node(values[i])
    queue.append(current.right)
    i+=1
print("Tree Created")
def inorder(root):
  if root:
    inorder(root.left)
    print(root.data,end=" ")
    inorder(root.right)
inorder(root)
```

```python
#create a tree and traverse the tree in postorder
class node:
  def __init__(self,data):
    self.data=data
    self.left=None
    self.right=None
values=input("enter elements:")
root=node(values[0])
queue=[root]
i=1
while i<len(values):
  current=queue.pop(0)
  if i<len(values):
    current.left=node(values[i])
    queue.append(current.left)
    i+=1
  if i<len(values):
    current.right=node(values[i])
    queue.append(current.right)
    i+=1
print("Tree Created")
def inorder(root):
  if root:
    print(root.data,end=' ')
    inorder(root.left)
    inorder(root.right)
inorder(root)
```
