```python
#check whether the parenthesis are balanced or not
s=input("Enter parenthesis:")
stack=[]
for ch in s:
  if ch=='(':
    stack.append(ch)
  elif ch==')':
    if len(stack)==0:
      print("Notbalanced")
      break
    stack.pop()
else:
  if len(stack)==0:
    print("balanced")
  else:
    print("not balanced")
```

```python
#adding the parenthesis if not balanced
s=input("Enter parenthesis:")
stack=[]
result=""
for ch in s:
  if ch=='(':
    stack.append(ch)
    result+=ch
  elif ch==')':
    if stack:
      stack.pop()
      result+=ch
    else:
      result='('+result+ch
while stack:
  result+=')'
  stack.pop()
print(result)
```

```python
#deleting parenthesis if not balanced
s=input("Enter parenthesis:")
stack=[]
result=[]
for ch in s:
  if ch=='(':
    stack.append(len(result))
    result.append(ch)
  elif ch==')':
    if stack:
      stack.pop()
      result.append(ch)
    else:
     continue
for i in reversed(stack):
  result.pop(i)
print("".join(result))
```
