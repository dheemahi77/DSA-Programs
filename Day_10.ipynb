**Searching** **Types**

```python
#linear search
arr=list(map(int,input("enter a elements:").split()))
target=int(input("enter a target:"))
for i in range(len(arr)):
  if arr[i]==target:
    print(target,"Found at index:",i)
    break
```

```python
#binary search
arr=list(map(int,input("enter a element:").split()))
l=0
r=len(arr)-1
target=int(input("enter a target:"))
found = False
while l <= r:
  m=(l+r)//2
  if arr[m]==target:
    print(target,"Found at index:",m)
    found = True
    break
  elif arr[m]<target:
    l=m+1
  else:
    r=m-1

if not found:
  print(target, "Not found in array")
```

```python
#jump search
import math
arr=list(map(int,input("enter elements:").split()))
key=int(input("enter a target:"))
n=len(arr)
step=int(math.sqrt(n))
prev=0
found=False
while prev<n and arr[min(step,n)-1]<key:
  prev=step
  step+=int(math.sqrt(n))
  if prev>=n:
    break
while prev<min(step,n):
  if arr[prev]==key:
    print("Element found at",prev)
    found=True
    break
  prev+=1
if not found:
  print("Element not found")
```
