```python
#code to read n values and traverse those values and print the counts of even numbers and count
n=int(input("enter size of array:"))
arr=[]
evencount,oddcount=0,0
for i in range(n):
  num=int(input("enter a value:"))
  arr.append(num)

for i in range(n):
  if arr[i]%2==0:
    evencount+=1
  else:
    oddcount+=1
print("evencount:",evencount)
print("oddcount:",oddcount)
```

```python
#code to read n values and traverse those values and print the +ve and -ve numbers
n=int(input("Enter size of array:"))
arr=[]
positivecount=0
negativecount=0
for i in range(n):
  num=int(input("Enter a value:"))
  arr.append(num)
for i in range(n):
  if arr[i]>0:
    positivecount+=1
  else:
    negativecount+=1
print("positivecount:",positivecount)
print("negativecount:",negativecount)
```
