```python
#write  a code to print the sum of digits of a given number
n=int(input("enter a number:"))
sum=0
while n!=0:
  d=n%10
  sum+=d
  n//=10
print("sum of digits",sum)
```

```python
#wirte a code to print reverse of a given number
n=int(input("enter a number:"))
rev=0
while n>0:
  d=n%10
  rev=rev*10+d
  n//=10
print("reverse number:",rev)
```

```python
#write a code to print count of even digits and odd digits of a given number
n=int(input("enter a number:"))
even=0
odd=0
while n!=0:
  d=n%10
  if d%2==0:
    even+=1
  else:
    odd+=1
  n//=10
print("even count",even)
print("odd count",odd)
```

```python
#anagram
a=input("enter 1 st string:").replace(' ','').lower()
b=input("enter 2 nd string:").replace(' ','').lower()
if len(a)!=len(b):
  print("Not anagram")
elif sorted(a)==sorted(b):
  print("Anagram")
else:
  print("not anagram")
```
